## 1. What if $\mathrm{CO}_2$ actually behaved like a Grey Gas?
____________

Suppose that $\mathrm{CO}_2$ actually behaved as a grey gas. In other words, no spectral dependence in absorptivity.

If we then **double the $\mathrm{CO}_2$ concentration** in the atmosphere, we double the number of absorbers. This should imply that we also **double the absorption cross-section** $\kappa$. 
<div style="color: green;">

To clarify, each molecule of $\mathrm{CO}_2$ always has the same absorption cross-section, but the absorption cross-section **per unit mass of air** increases if you assume that the fraction of $\mathrm{CO}_2$ in the air is small, and it does not contribute to the mass of the air. 

</div>

Following the notation from our [lecture on grey-gas radiative transfer](elementary-greenhouse#id-2-introducing-the-two-layer-grey-gas-model), we have:

$$ \kappa^\prime = 2 ~ \kappa $$

This would imply that we **double the optical thickness of every layer**:

$$ 
\Delta \tau^\prime  = 2 \left( -\frac{\kappa}{g} \Delta p \right) = 2 ~ \Delta \tau
$$ 

And since [the absorptivity / emissivity of each layer](radiative-transfer#emissivity) is

$$  \epsilon = 1 - \exp\big( - \Delta \tau \big)  $$

the **modified absorptivity** is

$$ \epsilon^\prime = 1 - \exp\big( - 2\Delta \tau \big) = 1 - \left( \exp\big( - \Delta \tau \big)\right)^2 = 1 - (1-\epsilon)^2 $$

or simply

$$ \epsilon^\prime = 2 \epsilon  - \epsilon^2 $$

(Note that $\epsilon^\prime = 2 \epsilon$ for very thin layers, for which $\epsilon$ is small).

### What does our 2-layer analytical model then say about the radiative forcing?

Recall that [we tuned the two-layer grey gas model](elementary-greenhouse#id-4-tuning-the-grey-gas-model-to-observations) with

$$ \epsilon = 0.586 $$

to get the observed OLR with observed temperatures.
<div style="color: green;">

$T_s = 288 \,\mathrm{K}, ~\mathrm{OLR} = 238.5 \,\mathrm{W}\,\mathrm{m}^{-2}$, and two air temperatures at different altitudes.

</div>


```python
#  Applying the above formula
eps = 0.586
print('Doubling a grey gas absorber would change the absorptivity from {:.3} to {:.3}'.format(eps, 2*eps - eps**2))
```

**If $\mathrm{CO}_2$ behaved like a grey gas**, doubling it would cause a huge increase in the absorptivity of each layer!

[We previously worked out that the radiative forcing in this model](elementary-greenhouse#the-radiative-forcing-change-in-olr-depends-on-the-lapse-rate) (with the observed lapse rate) is about +2.6 $\mathrm{W}\,\mathrm{m}^{-2}$ for a 2% increase in $\epsilon$.

<div style="color: green;">

$$ R = - \Delta \text{OLR} $$
$$
\begin{align*}
R &= -\Delta \text{OLR}_s - \Delta \text{OLR}_0 - \Delta \text{OLR}_1  \\
  &= -\Delta \epsilon \left[  -2(1-\epsilon) \sigma T_s^4 + (1 - 2 \epsilon) \sigma T_0^4  + \sigma T_1^4 \right]
\end{align*}
$$

</div>

```python
forcing_rate = 2.6 / 0.02   # W/m2 forcing per fractional increase in epsilon
```

Our hypothetical doubling of "grey $\mathrm{CO}_2$" yields an increase from 0.586 to 0.829, or more than 40%:

```python
fractional_increase = (0.829 - 0.586) / 0.586
print(fractional_increase)
```

So roughly speaking, this should yield a radiative forcing 20 times larger than +2.6 $\mathrm{W}\,\mathrm{m}^{-2}$:

```python
radiative_forcing = forcing_rate * fractional_increase  # W/m2
print(radiative_forcing)
```

**Our hypothetical doubling of "grey $\mathrm{CO}_2$" gives a radiative forcing greater than 50 $\mathrm{W}\,\mathrm{m}^{-2}$.** 

This is an absolutely enormous number. Assuming a [net climate feedback of -1.3 $\mathrm{W}\,\mathrm{m}^{-2}\,\mathrm{K}^{-1}$](sensitivity-feedback#id-7-feedbacks-diagnosed-from-complex-climate-models) (consistent with the CMIP5 model ensemble)
would then give us a truly astronomical **equilibrium climate sensitivity of 41 K**:
<div style="color: green;">

We haven't looked at this chapter yet, but climate feedback describes how the energy balance changes for each degree of change in surface temperature. It accounts for feedback effects such as water vapor feedback, surface albedo feedback (like we modelled in the last lecture), lapse rate feedback, etc. The equilibrium climate sensitivity is "the global mean surface warming necessary to balance the planetary energy budget after a doubling of atmospheric $\mathrm{CO}_2$".

</div>

```python
lambda_net = -1.3  #  W/m2/K
ecs = radiative_forcing / -(lambda_net)  # K
print(ecs)
```

It's time to move away from the Grey Gas approximation and look more carefully at the actual observed spectra of solar and terrestrial radiation.

<div style="color: green;">

![Solar spectrum](../images/solar_spectrum_2.png)

</div>

The sun produces light with a distribution similar to what would be expected from a 5525 K (5250 °C) blackbody, which is approximately the sun's surface temperature. As light passes through the atmosphere, some is absorbed by gases with specific absorption bands. Additional light is redistributed by Raleigh scattering, which is responsible for the atmosphere's blue color. - By the time the beam arrives at the surface, it is strongly depleted at specific wavelengths.
- Absorption by $\mathrm{O}_3$ (ozone) depletes almost the entire ultraviolet spectrum.
- Weaker absorption features, mostly due to $\mathrm{H}_2\mathrm{O}$, deplete some parts of the near-infrared.
- Note that the depletion in the visible band is mostly due to scattering, which depletes the direct beam but contributes diffuse radiation (so we can still see when it's cloudy!)

### Observed terrestrial spectra

This figure shows the Planck function for Earth's surface temperature compared with the spectrum observed from space.

![Terrestrial spectrum](../images/Terrestrial_spectrum.png)

Source: David Bice, Penn State University. https://www.e-education.psu.edu/earth103/node/1006

Careful: I'm pretty sure what is plotted here is not the **total** observed spectrum, but rather the part of the **emissions from the surface** that **actual make it out to space**.

As we now, the terrestrial beam from the surface is depleted by absorption by many greenhouse gases, but principally $\mathrm{CO}_2$ and $\mathrm{H}_2\mathrm{O}$.

However there is a spectral band centered on 10 $\mu\mathrm{m}$ in which the greenhouse effect is very weak. This is the so-called **window region** in the spectrum.

Since absorption is so strong across most of the rest of the infrared spectrum, this window region is a key determinant of the overall greenhouse effect.

#### One very big shortcoming of the Grey Gas model: it ignores the window region

We would therefore like to start using a model that includes enough spectral information that it represents

- the mostly strong $\mathrm{CO}_2$ absorption outside the window region
- the weak absorption inside the window region

## 3. Water vapor changes under global warming
____________

Another big shortcoming of the Grey Gas model is that it cannot represent the **water vapor feedback**.

We have seen above that $\mathrm{H}_2\mathrm{O}$ is an important absorber in both longwave and shortwave spectra.

We also know that the water vapor load in the atmosphere increases as the climate warms. The primary reason is that the **saturation vapor pressure** increases strongly with temperature. <div style="color:green;"> More on this later. </div>

<div style="color: green;">
Specific humidity: Mass of water vapor divided by total mass of moist air.

Relative humidity: How close the air is to being saturated with water vapor.

</div>

### What do you see here?

- Where does the largest warming occur?
- <div style="color: green;"> At the surface at the poles (ice melting). </div>
- Where does the largest moistening occur?
- <div style="color: green;"> At low altitudes near the equator, which is where temperatures are highest, and there is already lots of water vapor. </div>

In fact the specific humidity anomaly has roughly the same shape of the specific humidity field itself -- **it is largest where the temperature is highest**. This is a consequence of the Clausius-Clapeyron relation.

<div style="color: green;">

The Clausius-Clapeyron equation describes how a vapor's saturation pressure, $P_s$, changes with temperature.

$$
\frac{d(\ln{P_s})}{dT} = \frac{\Delta H_{\mathrm{vap}}}{RT^2}
$$

or

$$
\frac{dP_s}{dT} = \frac{\Delta H_{\mathrm{vap}}P_s}{RT^2}
$$

where $\Delta H_{\mathrm{vap}}$ is the enthalpy of vaporization and $R$ is the gas constant. We can see that the saturation pressure increases with temperature.


The equation for specific humidity is

$$
q=\frac{0.622e}{p-0.378e}
$$

where $e$ is the vapor partial pressure $e=RH\frac{P_s}{100} = H\,P_s$ ($RH$ is relative humidity), and $p$ is the total air pressure. $q$ increases with $e$, so if $P_s$ increases, $q$ increases. 

If we differentiate specific humidity with respect to temperature and assume $RH$ and $p$ stay constant, we get

$$
\frac{dq}{dT} = \frac{0.622H\,p}{(p-0.378H\,P_s)^2}\frac{dP_s}{dT}
$$

Substituting the Clausius-Clapeyron equation for $\frac{dP_s}{dT}$, and seeing that we have an expression for $q$ in the equation, we get

$$
\frac{dq}{dT} = q\frac{p}{p-0.378H\,P_s}\frac{\Delta H_{\mathrm{vap}}}{RT^2}
$$

We can see from this that the change in specific humidity will be larger where specific humidity was already large. Specific humidity will be larger where temperatures are higher, so his claim makes sense. 
</div>

The **relative humidity** anomaly is

- overall rather small (just a few percent)
- Largest in cold places where the specific humidity is very small.

The smallness of the relative humidity change is a rather remarkable result.

This is not something we can derive from first principles. It is an emergent property of the GCMs. However it is a very robust feature of global warming simulations.

## 4. A simple water vapor parameterization
____________

### A credible climate model needs a water vapor feedback

If relative humidity is nearly constant under global warming, and water vapor is a greenhouse gas, this implies a positive feedback that will amplify the warming for a given radiative forcing.

<div style="color:green;">

Increasing $T$ -> Increasing $P_s$ -> Increasing $q$ -> More water vapor -> More longwave absorption -> More energy staying in the system -> Increasing surface temperatures

</div>

Thus far our simple models have ignored this process, and we have not been able to use them to assess the climate sensitivity.

To proceed towards more realistic models, we have two options:

- **Simulate** all the evaporation, condensation and transport processes that determine the time-mean water vapor field (as is done in the CESM).
- **Parameterize** the dependence of water vapor on temperature by insisting that relative humidity stays constant as the climate changes.

We will now explore this second option, so that we can continue to think of the global energy budget under climate change as a process occurring in a single column.

### Manabe's constant relative humidity parameterization

We are going to adopt a parameterization first used in a very famous paper by {cite:t}`Manabe:1967aa`, *The thermal equilibrium of the atmosphere with a given distribution of relative humidity*.

This paper was the first to give a really credible calculation of climate sensitivity to a doubling of $\mathrm{CO}_2$ by accounting for the known spectral properties of $\mathrm{CO}_2$ and $\mathrm{H}_2\mathrm{O}$ absorption, as well as the water vapor feedback!

The parameterization is very simple:

We assume that the relative humidity $r$ is a linear function of pressure $p$:

$$ r = r_s \left( \frac{p/p_s - 0.02}{1 - 0.02} \right) $$

where $p_s = 1000$ hPa is the surface pressure, and $r_s$ is a prescribed surface value of relative humidity. Manabe and Wetherald set $r_s = 0.77$, but we should consider this a tunable parameter in our parameterization.

Since this formula gives a negative number above 20 hPa, we also assume that the **specific humidity** has a minimum value of $0.005$ g/kg (a typical stratospheric value).

This formula is implemented in `climlab.radiation.ManabeWaterVapor()`

Using this parameterization, the surface and tropospheric specific humidity will always increase as the temperature increases.

## 5. Modeling spectral bands with the `climlab.BandRCModel` process
____________

Here is a brief introduction to the `climlab.BandRCModel` process.

This is a model that divides the spectrum into 7 distinct bands: three shortwave and four longwave.

As we will see, the process works much like the familiar `climlab.RadiativeConvectiveModel`.

## About the spectra

![Solar spectrum](../images/solar_spectrum_2.png) ![Terrestrial spectrum](../images/Terrestrial_spectrum.png)

### Shortwave

The shortwave is divided into three channels:

- Channel 0 is the Hartley and Huggins band (extreme UV, 200 - 340 nm, 1% of total flux, strong ozone absorption)
- Channel 1 is Chappuis band (450 - 800 nm, 27% of total flux, moderate ozone absorption)
- Channel 2 is remaining radiation (72% of total flux, largely in the visible range, no ozone absorption)

![Solar absorption](../images/absorption_bands_1.png)


### Longwave

The longwave is divided into four bands:

- Band 0 is the **window region** (between 8.5 and 11 $\mu\mathrm{m}$), 17% of total flux.
- Band 1 is the $\mathrm{CO}_2$ absorption channel (the band of strong absorption by $\mathrm{CO}_2$ around 15 $\mu\mathrm{m}$), 15% of total flux
- Band 2 is a weak water vapor absorption channel, 35% of total flux
- Band 3 is a strong water vapor absorption channel, 33% of total flux

The longwave decomposition is not as easily related to specific wavelengths, as in reality there is a lot of overlap between $\mathrm{H}_2\mathrm{O}$ and $\mathrm{CO}_2$ absorption features (as well as absorption by other greenhouse gases such as $\mathrm{CH}_4$ and $\mathrm{N}_2\mathrm{O}$ that we are not representing).

### About the radiatively active gases

<div style="color:green;">

Remember that these gases will contribute to the emission and absorption of each atmospheric layer by affecting the optical thickness of the layer:

$$ \kappa(\nu, p, T) = \sum_{k=1}^n \kappa_k\big(\nu, p, T \big) q_k(p) $$

$$ \Delta \tau  = -\frac{\kappa}{g} \Delta p $$

$$ U_1 = U_0 ~ \exp(-\Delta \tau )  + E_{0} ~\Big( 1 - \exp(-\Delta \tau) \Big) $$

$$ D_0 = D_1 ~ \exp(-\Delta \tau )  + E_{0} ~\Big( 1 - \exp(-\Delta \tau) \Big) $$

The model performs these radiative transfers for each band, then sums up the total energy transfers to determine temperature changes at each timestep.
</div>

The Band model is aware of three different absorbing gases: O3 (ozone), CO2, and H2O (water vapor). The abundances of these gases are stored in a dictionary of arrays as follows:

```python
col1.absorber_vmr
```

Ozone and CO2 are both specified in the model. The default, as you see above, is zero ozone, and constant (well-mixed) CO2 at a volume mixing ratio of 3.8E-4 or 380 ppm.

Water vapor is handled differently: it is determined by the model at each timestep. We make the following assumptions, following {cite:t}`Manabe:1967aa`:

- the relative humidity just above the surface is fixed at 77% (can be changed of course... see the parameter `col1.relative_humidity`
- water vapor drops off linearly with pressure
- there is a small specified amount of water vapor in the stratosphere.

```python
#my addition#
h2o = col1.absorber_vmr['H2O']
fig, ax = plt.subplots()
ax.semilogy(h2o*1000, col1.lev)
ax.set_ylim(1000, col1.lev.min())
plt.xlabel("Specific humidity [g/kg]")
plt.ylabel("Pressure [hPa]")
plt.grid()
```

## Putting in some ozone

We need to provide some ozone data to the model in order to simulate a stratosphere. We will read in some ozone data just as we did in [the previous lecture on grey-gas modeling in climlab](grey-radiation-climlab#id-8-putting-stratospheric-ozone-in-the-grey-gas-model).

```python
ozone = xr.open_dataset( datapath + 'som_input/ozone_1.9x2.5_L26_2000clim_c091112.nc')
```

```python
#  Take global (area-weighted) and annual average
weight_ozone = cos(deg2rad(ozone.lat)) / cos(deg2rad(ozone.lat)).mean(dim='lat')
O3_global = (ozone.O3 * weight_ozone).mean(dim=('lat','lon','time'))
O3_global
```

```python
fig, ax = plt.subplots()
ax.plot( O3_global*1E6, ozone.lev)
ax.invert_yaxis()
ax.set_xlabel('Ozone (ppm)', fontsize=16)
ax.set_ylabel('Pressure (hPa)', fontsize=16 )
ax.set_title('Global, annual mean ozone concentration', fontsize = 16);
```

We are going to create another instance of the model, this time using the same vertical coordinates as the ozone data.

```python
#  Create the column with appropriate vertical coordinate, surface albedo and convective adjustment
col2 = climlab.BandRCModel(lev=ozone.lev)
col2
```

```python
#  Set the ozone mixing ratio
col2.absorber_vmr['O3'] = O3_global.values
```

```python
#my addition#
print(col2.ASR)

#  Run the model out to equilibrium!
col2.integrate_years(2.)

print(col2.ASR)
```

```python
fig, ax = plt.subplots()
ax.plot( col1.Tatm, np.log(col1.lev/1000), 'c-', label='RCE' )
ax.plot( col1.Ts, 0, 'co', markersize=16 )
ax.plot(col2.Tatm, np.log(col2.lev/1000), 'r-', label='RCE O3' )
ax.plot(col2.Ts, 0, 'ro', markersize=16 )
ax.invert_yaxis()
ax.set_xlabel('Temperature (K)', fontsize=16)
ax.set_ylabel('log(Pressure)', fontsize=16 )
ax.set_title('Temperature profiles', fontsize = 18)
ax.grid(); ax.legend()
```

Once we include ozone we get a well-defined stratosphere. 

Things to consider / try:

- Here we used the global annual mean Q = 341.3 $\mathrm{W}\,\mathrm{m}^{-2}$. We might want to consider latitudinal or seasonal variations in Q.
- We also used the global annual mean ozone profile! Ozone varies tremendously in latitude and by season. That information is all contained in the ozone data file we opened above. We might explore the effects of those variations.
- We can calculate climate sensitivity in this model by doubling the $\mathrm{CO}_2$ concentration and re-running out to the new equilibrium. Does the amount of ozone affect the climate sensitivity?  (example below)

```python
col3 = climlab.process_like(col2)
col3
```

```python
# Let's double CO2.
col3.absorber_vmr['CO2'] *= 2.
```

```python
col3.compute_diagnostics()
print( 'The radiative forcing for doubling CO2 is %f W/m2.' % 
      (col2.diagnostics['OLR'][0] - col3.diagnostics['OLR'][0]))
```

```python
col3.integrate_years(3)
```

```python
col3.ASR - col3.OLR
```

```python
print( 'The Equilibrium Climate Sensitivity is %f K.' % (col3.Ts[0] - col2.Ts[0]))
```

The current value of the surface relative humidity is 0.77. Let's see what happens when we scale the water vapor distribution and set $r_s=0.8$.

```python
col_more_h2o = climlab.process_like(col2)
col_more_h2o.subprocess["H2O"].relative_humidity
```

```python
col_more_h2o.subprocess["H2O"].relative_humidity = 0.8  #about 4% increase
col_more_h2o.subprocess["H2O"].RH_profile[:] = (col_more_h2o.subprocess["H2O"].relative_humidity * ((col_more_h2o.subprocess["H2O"].lev / const.ps - 0.02) / 0.98))
col_more_h2o.compute_diagnostics()
print('The radiative forcing for increasing H2O is %f W/m2.' % (col2.OLR[0] - col_more_h2o.OLR[0]))


col_more_h2o.integrate_years(3)
print(col_more_h2o.ASR - col_more_h2o.OLR)
print('The Equilibrium temperature response is %f K.' % (col_more_h2o.Ts[0] - col2.Ts[0]))
```
