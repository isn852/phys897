## Introduction

Consider the two-layer "leaky greenhouse" (or grey radiation) model from [these lecture notes](elementary-greenhouse).

Here you will use this model to investigate the **radiative effects of clouds**.

Clouds simultaneously **reflect shortwave radiation** and **absorb longwave radiation**. These two effects often oppose each other in nature, and which one is stronger depends (among other things) on whether the clouds are **low** or **high** (i.e. in layer 0 or layer 1).

For this question we will suppose (as we did in the lecture notes) that there is **no absorption of shortwave radiation** in the atmosphere.

## Question 1

Suppose a cloud reflects a fraction $\alpha_c$ of the shortwave beam incoming from above. $\alpha_c$ is a number between 0 and 1. Provide a coherent argument (in words, sketches, and/or equations) for why the **shortwave** effects of clouds should always be a **cooling** on the surface. Is this cooling effect different if the cloud is low or high? Explain.

### Q1 Answer

Considering only shortwave radiation, the leaky greenhouse model assumes that the atmosphere is transparent, and that a percentage of the incident insolation is absorbed by the surface, and the other fraction is reflected. The absorbed solar radiation is

\begin{equation}
\mathrm{ASR} = (1-\alpha)Q
\end{equation}

where $\alpha$ is the percentage of insolation that is reflected by the surface, or the albedo. If we consider additional reflection by clouds, then the amount of insolation absorbed by the surface is further reduced. If clouds reflect a fraction $\alpha_c$, then the new ASR is 

\begin{equation}
\mathrm{ASR'} = (1-\alpha_c)(1-\alpha)Q = (1-\alpha_c-\alpha+\alpha_c\alpha)Q  \tag{2}
\end{equation}

Since both $\alpha_c$ and $\alpha$ are less than 1, $\mathrm{ASR'}<\mathrm{ASR}$. We assumed that the atmosphere is transparent to shortwave radiation, so this reflected energy does not interact with any other parts of our model. If the Earth's surface absorbs less insolation, then it will cool, and so will atmospheric temperatures over time.


## Question 2

Because the liquid water droplets in a cloud are effective absorbers of longwave radiation, a cloud will **increase the longwave absorptivity / emissivity** of the layer in which it resides. 

We can represent this in the two-layer atmosphere by letting the absorptivity of a cloudy layer be $\epsilon + \epsilon_c$, where $\epsilon_c$ is an additional absorptivity due to the cloud. Derive a formula (i.e. an algebraic expression) for the OLR in terms of the temperatures $T_s, T_0, T_1$ and the emissivities $\epsilon, \epsilon_c$ for two different cases:

- a low cloud (the additional $\epsilon_c$ is in layer 0)
- a high cloud (the additional $\epsilon_c$ is in layer 1)

:::{hint}
Start from the formula from the lecture notes for OLR with separate emissivities for each layer $\epsilon_0, \epsilon_1$. 

The emissivity for layer 0 will either be $\epsilon_0 = \bar{\epsilon}$ or $\epsilon_0 = \bar{\epsilon} + \epsilon_c$ depending on where the cloud is, where $\bar{\epsilon}$ is the normal cloud-free emissivity which we assume to be the same in both layers.
:::

### Q2 Answers

Starting at the surface, the upward flux is

\begin{equation}
U_0 = \sigma T_s^4    \tag{4}
\end{equation}

Now at layer 0, a fraction of $U_0$ is absorbed, according to the layer's absorptivity, and the remaining fraction continues upwards. The layer also emits some radiation according to its emissivity. Let's call this layer's absorptivity/emissivity $\epsilon_0'$. The radiation that travels from layer 0 up to layer 1 is

\begin{equation}
U_1 = (1-\epsilon_0')U_0 + \epsilon_0' \sigma T_0^4    \tag{5}
\end{equation}

Finally, at the last layer, we have a similar situation, with this layer having $\epsilon_1'$:

\begin{equation}
U_2 = OLR = (1-\epsilon_1')U_1 + \epsilon_1' \sigma T_1^4    \tag{6}
\end{equation}

Substituing our expressions for $U_0$ and $U_1$, we find

\begin{equation}
OLR = (1-\epsilon_1')(1-\epsilon_0')\sigma T_s^4 + (1-\epsilon_1')\epsilon_0' \sigma T_0^4 + \epsilon_1' \sigma T_1^4    \tag{7}
\end{equation}

For the case of clouds in layer 0, $\epsilon_0' = \bar{\epsilon} + \epsilon_c$, and $\epsilon_1' = \bar{\epsilon}$. This would give 

\begin{equation}
 OLR = (1-\bar{\epsilon})(1-(\bar{\epsilon} + \epsilon_c))\sigma T_s^4 + (1-\bar{\epsilon})(\bar{\epsilon} + \epsilon_c) \sigma T_0^4 + \bar{\epsilon} \sigma T_1^4   \tag{8}
\end{equation}

If clouds are in layer 1, $\epsilon_0' = \bar{\epsilon}$, and $\epsilon_1' = \bar{\epsilon} + \epsilon_c$. We would then have

\begin{equation}
 OLR = (1-(\bar{\epsilon} + \epsilon_c))(1-\bar{\epsilon})\sigma T_s^4 + (1-(\bar{\epsilon} + \epsilon_c))\bar{\epsilon}\sigma T_0^4 + (\bar{\epsilon} + \epsilon_c) \sigma T_1^4   \tag{9}
\end{equation}

## Question 3

Now use the tuned numerical values we used in class:

- $T_s = 288$ K
- $T_0 = 275$ K
- $T_1 = 230$ K
- $\epsilon = 0.586$

and take $\epsilon_c = 0.2$

(a) Repeat the following for both a high cloud and a low cloud:

- Calculate the **difference in OLR** due to the presence of the cloud, compared to the case with no cloud. 
- Does this represent a warming or cooling effect?

(b) Which one has a larger effect, the low cloud or the high cloud?

```python
T_s = 288
T_0 = 275
T_1 = 230
epsilon = 0.586
epsilon_c = 0.2
sigma = 5.67E-8

def OLR(T_s, T_0, T_1, epsilon, epsilon_c, cloud_layer):
    if cloud_layer == 0:
        epsilon_0 = epsilon + epsilon_c
        epsilon_1 = epsilon
    elif cloud_layer == 1:
        epsilon_0 = epsilon
        epsilon_1 = epsilon + epsilon_c

    return (1-epsilon_1)*(1-epsilon_0)*sigma*T_s**4 + (1-epsilon_1)*epsilon_0*sigma*T_0**4 + epsilon_1*sigma*T_1**4


## no cloud ##
epsilon_c = 0
OLR_no_cloud = OLR(T_s=T_s, T_0=T_0, T_1=T_1, epsilon=epsilon, epsilon_c=epsilon_c, cloud_layer=0)

## high cloud ## 
cloud_layer = 1
epsilon_c = 0.2
OLR_high_cloud = OLR(T_s=T_s, T_0=T_0, T_1=T_1, epsilon=epsilon, epsilon_c=epsilon_c, cloud_layer=cloud_layer)

## low cloud ## 
cloud_layer = 0
epsilon_c = 0.2
OLR_low_cloud = OLR(T_s=T_s, T_0=T_0, T_1=T_1, epsilon=epsilon, epsilon_c=epsilon_c, cloud_layer=cloud_layer)

print("high_cloud - no_cloud = %0.2f W/m2" %(OLR_high_cloud-OLR_no_cloud))
print("low_cloud - no_cloud = %0.2f W/m2" %(OLR_low_cloud-OLR_no_cloud))
```

In both cases, the new OLR after the addition of clouds is less than the OLR with no clouds. This means that more energy is staying in the Earth system, and temperatures will increase to bring the system into equillibrium. The addition of clouds into the higher layer had a stronger effect on the OLR, reducing it over 7 times more than the addition of clouds into the lower layer.

## Question 5

How would your answer change if the atmosphere were **isothermal**, i.e. $T_s = T_0 = T_1$?

Similar to our findings without clouds, an isothermal atmosphere does not change in response to variations in emissivity/absorptivity due to the presence of clouds. The shortwave effect would still be present, so the addition of clouds at any height in an iso thermal atmosphere would cause cooling.
